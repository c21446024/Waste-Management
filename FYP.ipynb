# Importing and Reading 2023 Vehicles


```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import pandas as pd
import numpy as np
import os
```

```python
vehicles_2023 = pd.read_csv('/content/drive/MyDrive/FYP/Collisions/dft-road-casualty-statistics-collision-2023 (3).csv')
```

```python
df = vehicles_2023
```

```python
print("Dataset Information:")
print(df.info())
```

```python
print("\nSummary Statistics:")
print(df.describe())
```

```python
print("\nMissing Values Per Column:") # check missing values for preprocessing
print(df.isnull().sum())
```

```python
# Replace -1 values with NaN
df.replace(-1, np.nan, inplace=True)

# Dropping columns with more than 50% missing values
threshold = 0.5  # 50% threshold
missing_fraction = df.isna().mean()
cols_to_drop = missing_fraction[missing_fraction > threshold].index
df.drop(columns=cols_to_drop, inplace=True)

print(f"Dropped columns with >50% missing data: {list(cols_to_drop)}")

# Fill in the remaining NaN values in numeric columns with the median
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

folder_path = '/content/drive/MyDrive/FYP/Clean_Data'



file_path = os.path.join(folder_path, 'cleaned_vehicles_2023.csv')
df.to_csv(file_path, index=False)
print(f" File saved at: {file_path}")
```

```python
df['accident_hour'] = pd.to_datetime(df['time'], format='%H:%M', errors='coerce').dt.hour

# Group hours into time of day
def map_time_of_day(hour):
    if 5 <= hour <= 11:
        return 'Morning'
    elif 12 <= hour <= 16:
        return 'Afternoon'
    elif 17 <= hour <= 20:
        return 'Evening'
    else:
        return 'Night'

df['Time_of_Day'] = df['accident_hour'].apply(map_time_of_day)

plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    x='Time_of_Day',
    hue='Accident_Severity_Label',   # Group by severity
    palette="Set2",
    order=["Morning", "Afternoon", "Evening", "Night"]
)

plt.title("Accidents by Time of Day and Severity", fontsize=18)
plt.xlabel("Time of Day", fontsize=14)
plt.ylabel("Number of Accidents", fontsize=14)
plt.legend(title="Accident Severity")
plt.tight_layout()

# Save
save_folder = "/content/drive/MyDrive/FYP/plots"
os.makedirs(save_folder, exist_ok=True)
save_path = os.path.join(save_folder, "accidents_by_time_of_day_severity.png")
plt.savefig(save_path, dpi=300)

plt.show()

print(f"Chart saved to: {save_path}")
```

# Mapping To Full Address

```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut
import time
from tqdm import tqdm
import re
```

```python
# Initialise geolocator
geolocator = Nominatim(user_agent="accident_analysis")

# Load CSV in chunks since its crashes when i run it all at one
chunk_size = 2000  # Process 2000 rows at a time
input_file = "long_lat.csv"
output_file = "accident_data_with_locations.csv"

# Dictionary to store cached results
location_cache = {}

def reverse_geocode(lat, lon):
    """Reverse geocode function with caching"""
    coords = (round(lat, 4), round(lon, 4))
    if coords in location_cache:
        return location_cache[coords]

    try:
        location = geolocator.reverse(coords, language="en", timeout=10)
        if location:
            address = location.raw.get("address", {})
            result = {
                "full_address": location.address,
                "street": address.get("road", "N/A"),
                "city": address.get("city", address.get("town", address.get("village", "N/A"))),
                "suburb": address.get("suburb", "N/A"),
                "state": address.get("state", "N/A"),
                "country": address.get("country", "N/A"),
            }
            location_cache[coords] = result
            return result
        else:
            return {"full_address": None, "street": None, "city": None, "suburb": None, "state": None, "country": None}

    except GeocoderTimedOut:
        time.sleep(1)  # Retry after a short delay
        return reverse_geocode(lat, lon)
    except Exception as e:
        print(f"Error: {e}")
        return {"full_address": None, "street": None, "city": None, "suburb": None, "state": None, "country": None}

# Process in chunks
first_chunk = True  # To handle CSV headers

for chunk in pd.read_csv(input_file, chunksize=chunk_size):
    tqdm.pandas()

    # Apply geocoding to the chunk
    chunk[["full_address", "street", "city", "suburb", "state", "country"]] = chunk.progress_apply(
        lambda row: pd.Series(reverse_geocode(row["latitude"], row["longitude"])), axis=1
    )

    # Apdd to CSV
    chunk.to_csv(output_file, mode="w" if first_chunk else "a", header=first_chunk, index=False)
    first_chunk = False  # Ensures headers are written only once

    print(f"Processed {chunk_size} rows, saved to {output_file}")

print("Reverse geocoding completed. File saved.")
```

```python
df = pd.read_csv("/content/long_lat.csv")

df = df.dropna(subset=["longitude", "latitude"])
coords = df[["longitude", "latitude"]]
scaler = StandardScaler()
coords_scaled = scaler.fit_transform(coords)

# Compute inertia for k from 2 to 20
inertia = []
k_range = range(2, 21)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(coords_scaled)
    inertia.append(kmeans.inertia_)


plt.figure(figsize=(10, 6))
plt.plot(k_range, inertia, marker='o', label='Inertia')
plt.axvline(x=15, color='red', linestyle='--', label='Chosen k = 15') # marker at 15
plt.title("Elbow Method for Determining Optimal k")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia (Within-Cluster Sum of Squares)")
plt.grid(True)
plt.legend()
plt.tight_layout()

# Save the plot
output_marked = "/content/drive/MyDrive/FYP/elbow_plot_kmeans_marked.png"
plt.savefig(output_marked, dpi=300)
plt.show()

output_marked
```

```python
import pandas as pd
import geopandas as gpd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import contextily as ctx
from shapely.geometry import Point

df = pd.read_csv("/content/long_lat.csv")
df = df.dropna(subset=["longitude", "latitude"])

# Convert to GeoDataFrame
geometry = [Point(xy) for xy in zip(df["longitude"], df["latitude"])]
gdf = gpd.GeoDataFrame(df, geometry=geometry, crs="EPSG:4326")

# Project to Web Mercator for plotting with basemap
gdf = gdf.to_crs(epsg=3857)

scaler = StandardScaler()
coords_scaled = scaler.fit_transform(df[["longitude", "latitude"]])
kmeans = KMeans(n_clusters=15, random_state=42)
gdf["cluster"] = kmeans.fit_predict(coords_scaled)

fig, ax = plt.subplots(figsize=(12, 10))
gdf.plot(ax=ax, column="cluster", cmap="tab20", markersize=5, legend=True)

# Add basemap of the UK
ctx.add_basemap(ax, source=ctx.providers.OpenStreetMap.Mapnik)
ax.set_title("UK Traffic Coordinates Clustered (KMeans with Basemap)")
plt.axis("off")
plt.tight_layout()
plt.savefig("cluster_plot_with_basemap.png", dpi=300)
plt.show()
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import seaborn as sns
import matplotlib.pyplot as plt

# Step 1: Load your binary dataset
df = pd.read_csv('/content/drive/MyDrive/FYP/merged_binary_severity.csv')

# Step 2: Sample to avoid memory errors
df_sampled = df.dropna(subset=['binary_severity']).sample(n=10000, random_state=42)

# Step 3: Select features (you can use correlation/importance list again)
features = [
    'latitude',
    'location_northing_osgr',
    'longitude',
    'location_easting_osgr',
    'age_of_driver',
    'make_freq',
    'engine_capacity_cc',
    'did_police_officer_attend_scene_of_accident',
    'vehicle_manoeuvre',
    'speed_limit'
]

X = df_sampled[features].fillna(df_sampled.median(numeric_only=True))
y = df_sampled['binary_severity']

# Step 4: Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Step 5: Train model
rf_binary = RandomForestClassifier(n_estimators=100, random_state=42)
rf_binary.fit(X_train, y_train)

# Step 6: Predict and evaluate
y_pred = rf_binary.predict(X_test)

print("✅ Accuracy:", accuracy_score(y_test, y_pred))
print("\n📊 Classification Report:\n", classification_report(y_test, y_pred))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix - Random Forest (Binary Severity, No SMOTE)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.savefig("correlations/conf_matrix_binary_no_smote.png", dpi=300)
plt.show()
```

```python

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
import seaborn as sns
import matplotlib.pyplot as plt
from xgboost import XGBClassifier, plot_importance


df = pd.read_csv('/content/drive/MyDrive/FYP/Vehicles_Collisions2023')

df = df.drop(columns=[
    'accident_index_x', 'accident_year_x', 'accident_reference',
    'lsoa_of_driver', 'generic_make_model', 'driver_home_area_type',
    'dir_from_e', 'dir_from_n', 'dir_to_e', 'dir_to_n'
], errors='ignore')

df = df.dropna()

for col in df.select_dtypes(include=['object', 'category']).columns:
    df.loc[:, col] = df[col].astype(str)
    df.loc[:, col] = LabelEncoder().fit_transform(df[col])


target = 'accident_severity'
X = df.drop(columns=['accident_severity'])
X = X.select_dtypes(include=[np.number])

y = df['accident_severity'] - 1

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

from collections import Counter
counter = Counter(y_train)
print("Class distribution:", counter)
majority = max(counter.values())
weights = {cls: float(majority)/count for cls, count in counter.items()}
print("Class weights:", weights)

model = XGBClassifier(
    objective='multi:softmax',
    num_class=3,
    eval_metric='mlogloss',
    use_label_encoder=False,
    random_state=42,
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    scale_pos_weight=weights[1]  # Adjusting for minority class (class 1)
)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(" Accuracy Score:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

plt.figure(figsize=(6, 4))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from imblearn.over_sampling import SMOTE
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("/content/drive/MyDrive/FYP/Collisions/dft-road-casualty-statistics-collision-2023 (3).csv")

# Drop rows with missing or invalid target
df = df[df['enhanced_severity_collision'] != -1]

# Define features and target (add more features as needed)
features = ['number_of_casualties', 'number_of_vehicles', 'speed_limit']
X = df[features]
y = df['enhanced_severity_collision'].astype('category').cat.codes  # Encode labels

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Apply SMOTE to training data only
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_scaled, y_train)

# Train Random Forest
rf_model = RandomForestClassifier(class_weight='balanced', random_state=42)
rf_model.fit(X_train_resampled, y_train_resampled)

# Predict on test set
y_pred = rf_model.predict(X_test_scaled)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 5))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix - Random Forest with SMOTE")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = pd.read_csv("/content/drive/MyDrive/FYP/merged_vehicles_collisions_cleaned.csv", low_memory=False)
df = df.dropna()

features = [
    'latitude',
    'location_northing_osgr',
    'longitude',
    'location_easting_osgr',
    'age_of_driver',
    'make_freq',
    'engine_capacity_cc',
    'did_police_officer_attend_scene_of_accident',
    'vehicle_manoeuvre',
    'speed_limit'
]

X = df[features]
y = df['accident_severity']  # original labels: 1, 2, 3

# Encode categorical features
for col in X.columns:
    if X[col].dtype == 'object':
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col])

# Convert to zero-indexed labels for training
y_encoded = y - 1

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.3, random_state=42, stratify=y_encoded
)

# Train XGBoost
xgb = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    use_label_encoder=False,
    eval_metric='mlogloss'
)
xgb.fit(X_train, y_train)

# Predict
y_pred = xgb.predict(X_test)

# Adjust predictions back to 1, 2, 3
y_pred_adjusted = y_pred + 1
y_test_adjusted = y_test + 1

# Evaluate
print(" Accuracy:", accuracy_score(y_test_adjusted, y_pred_adjusted))
print("\n Classification Report:\n", classification_report(y_test_adjusted, y_pred_adjusted))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test_adjusted, y_pred_adjusted)

# Set2-style green tone for heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap=sns.light_palette("#8da0cb", as_cmap=True))
plt.title("Confusion Matrix - XGBoost top 10 features of importance", fontsize=13)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()

# Save the plot
save_folder = "/content/drive/MyDrive/FYP/cofusion_matrix"
os.makedirs(save_folder, exist_ok=True)
save_path = os.path.join(save_folder, "xgboost_import_features_1.png")
plt.savefig(save_path, dpi=300)
plt.show()

print(f" Confusion matrix saved to: {save_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical
from imblearn.over_sampling import SMOTE
import os

# Load data
df = pd.read_csv("/content/drive/MyDrive/FYP/merged_vehicles_collisions_cleaned.csv")

df['accident_severity'] = df['accident_severity'] - 1  # Shift to 0,1,2

# Features
features = [
    'latitude',
    'location_northing_osgr',
    'longitude',
    'location_easting_osgr',
    'age_of_driver',
    'make_freq',
    'engine_capacity_cc',
    'did_police_officer_attend_scene_of_accident',
    'vehicle_manoeuvre',
    'speed_limit',
    # 'cluster'  # Add if using cluster versions
]

X = df[features]
y = df['accident_severity']

# Apply SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)
print("After SMOTE:", np.bincount(y_resampled))

# One-hot encode labels
y_resampled_cat = to_categorical(y_resampled)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled_cat, test_size=0.2, random_state=42)

# Feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Build MLP model
model = Sequential()
model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))
model.add(Dropout(0.3))
model.add(Dense(32, activation='relu'))
model.add(Dense(3, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
early_stop = EarlyStopping(monitor='val_loss', patience=3)

# Train model
history = model.fit(X_train, y_train, validation_split=0.2, epochs=100, batch_size=32,
                    callbacks=[early_stop], verbose=1)

# Predict
y_pred_prob = model.predict(X_test)
y_pred = np.argmax(y_pred_prob, axis=1)
y_true = np.argmax(y_test, axis=1)

# Classification Report
print(classification_report(y_true, y_pred, target_names=['Fatal', 'Serious', 'Slight']))

# === Save Confusion Matrix ===
conf_matrix = confusion_matrix(y_true, y_pred)

# Plot and save confusion matrix
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap="Purples", xticklabels=['Fatal', 'Serious', 'Slight'], yticklabels=['Fatal', 'Serious', 'Slight'])
plt.title("Confusion Matrix - MLP (Important Features)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()

# Save image
save_folder = "/content/drive/MyDrive/FYP/MLP"
os.makedirs(save_folder, exist_ok=True)
save_path = os.path.join(save_folder, "mlp_important_features_smote.png")
plt.savefig(save_path, dpi=300)
plt.show()

print(f"Confusion matrix saved to: {save_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from imblearn.over_sampling import SMOTE
import os

# Load data
df = pd.read_csv("/content/drive/MyDrive/FYP/merged_binary_severity_clusters.csv")

# Features
features = [
    'latitude',
    'location_northing_osgr',
    'longitude',
    'location_easting_osgr',
    'age_of_driver',
    'make_freq',
    'engine_capacity_cc',
    'did_police_officer_attend_scene_of_accident',
    'vehicle_manoeuvre',
    'speed_limit',
    # 'cluster'  # Uncomment if using cluster version
]

X = df[features].fillna(df[features].mean())
y = df['binary_severity']  # Use binary target (0 = Non-Severe, 1 = Severe)

# Apply SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)
print("After SMOTE:", np.bincount(y_resampled))

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, test_size=0.2, random_state=42, stratify=y_resampled
)

# Feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Build binary MLP model
model = Sequential()
model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))
model.add(Dropout(0.3))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='sigmoid'))  # One output neuron with sigmoid for binary classification

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
early_stop = EarlyStopping(monitor='val_loss', patience=3)

# Train model
history = model.fit(X_train, y_train, validation_split=0.2, epochs=100, batch_size=32,
                    callbacks=[early_stop], verbose=1)

# Predict
y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob > 0.5).astype("int").flatten()  # Convert probabilities to binary predictions

# Classification Report
print(classification_report(y_test, y_pred, target_names=['Non-Severe', 'Severe']))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap="Purples",
            xticklabels=['Non-Severe', 'Severe'],
            yticklabels=['Non-Severe', 'Severe'])
plt.title("Confusion Matrix - MLP (Binary Severity - Important Features)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()

# Save image
save_folder = "/content/drive/MyDrive/FYP/MLP"
os.makedirs(save_folder, exist_ok=True)
save_path = os.path.join(save_folder, "mlp_binary_severity_important_features.png")
plt.savefig(save_path, dpi=300)
plt.show()

print(f"Confusion matrix saved to: {save_path}")
```

```python
# Install keras-tuner if you haven't
!pip install keras-tuner --quiet

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import keras_tuner as kt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical
from imblearn.over_sampling import SMOTE
import os

# Load and prepare data
df = pd.read_csv("/content/drive/MyDrive/FYP/Clean_Data/v_c_merged_with_clusters.csv")
df['accident_severity'] = df['accident_severity'] - 1
df['cluster'] = df['cluster'].fillna(df['cluster'].mean())

features = [
    'latitude',
    'location_northing_osgr',
    'longitude',
    'location_easting_osgr',
    'age_of_driver',
    'make_freq',
    'engine_capacity_cc',
    'did_police_officer_attend_scene_of_accident',
    'vehicle_manoeuvre',
    'speed_limit',
    'cluster'
]

X = df[features].fillna(df[features].mean())
y = df['accident_severity']

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)
y_resampled_cat = to_categorical(y_resampled)

X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled_cat, test_size=0.2, random_state=42
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Define model builder function for Keras Tuner
def build_model(hp):
    model = Sequential()
    model.add(Dense(
        units=hp.Int('units_input', min_value=32, max_value=128, step=32),
        activation='relu',
        input_dim=X_train.shape[1]
    ))
    model.add(Dropout(hp.Float('dropout_input', min_value=0.2, max_value=0.5, step=0.1)))

    # Optional second layer
    if hp.Boolean('second_layer'):
        model.add(Dense(
            units=hp.Int('units_hidden', min_value=32, max_value=128, step=32),
            activation='relu'
        ))
        model.add(Dropout(hp.Float('dropout_hidden', min_value=0.2, max_value=0.5, step=0.1)))

    model.add(Dense(3, activation='softmax'))  # 3 classes

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

# Initialize the tuner
tuner = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=15,  # Try 15 random models
    executions_per_trial=1,
    directory='/content/drive/MyDrive/FYP/MLP_Tuning',
    project_name='mlp_hyperparameter_tuning'
)

# Early stopping
early_stop = EarlyStopping(monitor='val_loss', patience=3)

# Start the search
tuner.search(X_train, y_train, epochs=50, validation_split=0.2, callbacks=[early_stop], verbose=1)

# Get the best model
best_model = tuner.get_best_models(num_models=1)[0]

# Predict
y_pred_prob = best_model.predict(X_test)
y_pred = np.argmax(y_pred_prob, axis=1)
y_true = np.argmax(y_test, axis=1)

# Evaluation
print("\n Classification Report:\n", classification_report(y_true, y_pred, target_names=['Fatal', 'Serious', 'Slight']))

conf_matrix = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap="Purples",
            xticklabels=['Fatal', 'Serious', 'Slight'],
            yticklabels=['Fatal', 'Serious', 'Slight'])
plt.title("Confusion Matrix - Tuned MLP (Important Features + Clusters)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()

# Save confusion matrix
save_folder = "/content/drive/MyDrive/FYP/MLP"
os.makedirs(save_folder, exist_ok=True)
save_path = os.path.join(save_folder, "mlp_tuned_important_features_cluster.png")
plt.savefig(save_path, dpi=300)
plt.show()

print(f"Confusion matrix saved to: {save_path}")
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb
import lightgbm as lgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from imblearn.over_sampling import SMOTE
import pandas as pd
import numpy as np
import os
import seaborn as sns
import matplotlib.pyplot as plt

#Load dataset
df = pd.read_csv('/content/drive/MyDrive/FYP/Clean_Data/v_c_merged_with_clusters.csv')
df = df.dropna(subset=['accident_severity'])

features = [
    'latitude',
    'location_northing_osgr',
    'longitude',
    'location_easting_osgr',
    'age_of_driver',
    'make_freq',
    'engine_capacity_cc',
    'did_police_officer_attend_scene_of_accident',
    'vehicle_manoeuvre',
    'speed_limit',
    'cluster'
]
X = df[features].fillna(df[features].median(numeric_only=True))
y = df['accident_severity'] - 1  # 0-indexed classes

# SMOTE balancing
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, test_size=0.3, random_state=42, stratify=y_resampled
)

# Define base models
base_models = [
    ('rf', RandomForestClassifier(n_estimators=100, random_state=42)),
    ('xgb', xgb.XGBClassifier(objective='multi:softmax', num_class=3, eval_metric='mlogloss', random_state=42, use_label_encoder=False)),
    ('lgbm', lgb.LGBMClassifier(objective='multiclass', num_class=3, random_state=42))
]

# Define meta-model
meta_model = LogisticRegression(max_iter=1000, multi_class='multinomial', solver='lbfgs')

# Build Stacking Classifier
stacking_clf = StackingClassifier(
    estimators=base_models,
    final_estimator=meta_model,
    cv=3,
    n_jobs=-1,
    passthrough=False
)

# Train stacking model
stacking_clf.fit(X_train, y_train)

#  Predict and evaluate
y_pred = stacking_clf.predict(X_test)
y_pred_adjusted = y_pred + 1
y_test_adjusted = y_test + 1

print(" Stacking Model Accuracy:", accuracy_score(y_test_adjusted, y_pred_adjusted))
print("\n Classification Report:\n", classification_report(y_test_adjusted, y_pred_adjusted))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test_adjusted, y_pred_adjusted)
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap="Purples",
            xticklabels=['Fatal', 'Serious', 'Slight'],
            yticklabels=['Fatal', 'Serious', 'Slight'])
plt.title("Confusion Matrix - Stacking Ensemble")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()

# Save confusion matrix
save_folder = "/content/drive/MyDrive/FYP/Stacking"
os.makedirs(save_folder, exist_ok=True)
save_path = os.path.join(save_folder, "stacking_confusion_matrix.png")
plt.savefig(save_path, dpi=300)
plt.show()

print(f" Confusion matrix saved to: {save_path}")
```
